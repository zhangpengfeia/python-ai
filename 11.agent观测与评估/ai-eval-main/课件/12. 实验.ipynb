```python
from langfuse import get_client

langfuse = get_client()
```

```python
page = 1
datasets = []
while True:
    result = langfuse.api.datasets.list(page=page, limit=100)
    datasets.extend(result.data)
    if page >= result.meta.total_pages:
        break
    page += 1

datasets
```

```python
from ai_eval.benchmarks.dataset import get_items

items = get_items('pre-build')
items
```

```python
from langfuse.api import DatasetItem

async def experiment_task(*, item:DatasetItem, **kwargs):
    return item.expected_output

# 运行实验

langfuse.run_experiment(
    # 给实验取个名字
    name="主Agent的回归实验",
    # 如果不配置，则使用name+当前时间自动生成
    run_name=None,
    description="描述1",
    # 此次实验针对哪些数据项
    data=items,
    task=experiment_task # type: ignore
)
```
