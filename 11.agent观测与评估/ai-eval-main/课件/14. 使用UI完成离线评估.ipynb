3. 根据关键内容的遗漏、错误或矛盾程度降低分数；关键结论相反时应给低分。
4. 不因合理的补充说明扣分，除非补充内容与期望输出冲突。
5. 只依据提供的两段内容评分，不自行补充事实。

- 0.25：只有少量内容一致，主要结论或要求未满足。
- 0.0：完全不符、关键结论相反，或实际输出为空。
