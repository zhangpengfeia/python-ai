# 资源隔离

## 普通资源隔离

思路如下：

- 对需要隔离的资源，在创建时标记一个隔离键，通常是用户ID。
- 对资源的后续操作，需要用隔离键进行筛选。

```python
from langgraph_sdk import get_client

user1 = get_client(
    url="http://127.0.0.1:2024",
    headers={"Authorization": "Bearer user_1"}
)

user2 = get_client(
    url="http://127.0.0.1:2024",
    headers={"Authorization": "Bearer user_2"}
)

assistant_id = 'c3cdd36b-80ed-5994-823a-a26808ae54cd'
```

```python
# 创建新线程
thread = await user1.threads.create(
    metadata={"__name__": "用户1的线程3"}
)
thread_id = thread["thread_id"]
thread
```

```python
# 查找线程
await user2.threads.search()
```

## Store隔离

Store不能依靠隔离键处理，需要使用命名空间进行隔离

- 对于用户自由存储的资源，必须以正确的用户ID作为命名空间前缀
- 对于图代码、后端代码等资源，自行使用命名空前前缀隔离

```python
await user1.store.put_item(
    ["user_id_001", "store-demo"],
    key="preference",
    value={
        "theme":"dark",
        "lang": "zh-cn"
    }
)
```

```python
await user1.store.search_items(
    ["user_id_001", "store-demo"]
)
```
