Agent Server 的鉴权功能决定了不同的用户可以执行什么操作。

```python
from langgraph_sdk import get_client

client = get_client(url="http://127.0.0.1:2024")
```

```python
# 创建 Assistant
await client.assistants.create(
    graph_id="main_agent",
    name="主代理", 
    description="最多允许运行100个超步",
    config={
        "recursion_limit": 100
    },
    headers={
        "Authorization": "Bearer admin"
    }
)
```
